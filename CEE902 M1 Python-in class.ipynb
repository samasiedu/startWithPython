## plotting

### More examples (more controls and other types of plots) using the matplotlib can be found at:

```python
# install a library if not already installed using Anaconda or the code below
# %pip install numpy

# import the matlab plotting libray 
# import the numpy library
import matplotlib.pylab as plt 
import numpy as np


#use the numpy linspace function to create a numpy array strating with 0 and ending with 10, with a
#total number of 15 numbers 
#You can play with this number to make the curve smoother
x = np.linspace(0,10,15)

#x a numpy array, you can use type() function to see the type of a variable, e.g.,
print("type for x is:",type(x))

#you can also check the dimension of the array:
print("shape for x is:", x.shape)

print(x)

#y is sin(x), here y takes the shape/size of x
y=np.sin(x)
print(y)

#plot the x vs. y curve
plt.plot(x,y)


```
