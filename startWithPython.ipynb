```python
import numpy as np
import matplotlib.pyplot as plt

# read data from file
xdata, ydata = np.loadtxt('wavyPulseData.txt', unpack=True)

x = np.linspace(-10., 10., 200)
y = np.sin(x) * np.exp(-(x/5.0)**2)

# creating plot
plt.figure(1, figsize=(6, 4))
plt.plot(x, y, 'g-', label='theory')        # g- = green liine
plt.plot(xdata, ydata, 'bo', label='data')  #bo = blue circles
plt.xlabel('x')
plt.ylabel('transverse displacement')
plt.legend(loc='upper right', title='legend')
plt.axhline(color='gray', zorder=-1)
plt.axvline(color='gray', zorder=-1)

plt.savefig('WavyPulse.png')

plt.show()
```
